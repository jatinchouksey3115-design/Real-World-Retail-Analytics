```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("../data/retail_transactions_raw.csv")
display(df.head())
```

## Data Understanding and Preparation


```python
display(df.info())
display(df.describe(include="all").T)
print(df.isna().sum())
print("Duplicates:",df.duplicated().sum())
df["Order_Date"]=pd.to_datetime(df["Order_Date"])
df["Customer_ID"]=df["Customer_ID"].fillna("Unknown")
df=df.drop_duplicates()
```

## Core Retail KPIs


```python
print("Revenue: ₹{:,.2f}".format(df.Sales.sum()))
print("Orders:",df.Order_ID.nunique())
print("Unique customers:",df.loc[df.Customer_ID!="Unknown","Customer_ID"].nunique())
print("Average order value: ₹{:,.2f}".format(df.Sales.sum()/df.Order_ID.nunique()))
```
